```python
result = 0

def add(num):
    global result
    result += num
    return result

print(add(5))
print(add(10))
print(add(3))
```

```python
result2 = 0

def add2(num):
    global result2
    result2 += num
    return result2

print(add2(3))
print(add2(5))
```

```python
class calculator:
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result


cal1 = calculator()
print(cal1.add(5))
print(cal1.add(10))

cal2 = calculator()
print(cal2.add(3))
print(cal2.add(8))
```

```python
# 속성 사용하기

# class 클래스이름:
#     def __init__(self):
#         self.속성명 = 값

class person:
    def __init__(self):
        self.hello = "안녕하세요"
    
    def greeting(self):
        print(self.hello)
        
james = person()
print(james.hello)
james.greeting()
```

```python
# 인스턴스를 만들때 값 받기

# class 클래스이름:
#     def __init__(self,매개변수1, 매개변수2):
#         self.속성1 = 매개변수1
#         self.속성2 = 매개변수2

class person:
    def __init__(self, name, age, address):
        self.hello = "안녕하세요"
        self.name = name
        self.age = age
        self.address = address
        
    def greeting(self):
        print("{0} 저는 {1}입니다.".format(self.hello, self.name))
        
maria = person("마리아", 20, "서울시 서초구 반포동")
maria.greeting()

print("이름 : ", maria.name)
print("나이 : ", maria.age)
print("주소 : ", maria.address)
```
