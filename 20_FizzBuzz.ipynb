```python
# 3의 배수일때 Fizz 5의 배수일때 Buzz
## 3과 5의 공배수 15는 처음 피즈에서 걸러져서 피즈로 나옴

for i in range(1, 21):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 이것도 처음 피즈에 걸려져서 그 밑에 있는 피즈버즈가 나올수가 없음

for i in range(1, 21):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 == 0 and i % 5 == 0: # i % 15 == 0
        print("FizzBuzz")
    else:
        print(i)
```

```python
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0: # i % 15 == 0
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기

for i in range(1, 31):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or i)
```
