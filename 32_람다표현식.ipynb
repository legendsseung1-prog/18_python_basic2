```python
# 람다 표현식으로 함수 만들기

def plus_ten(x):
    return x + 10

plus_ten(1)
```

```python
# lambda 매개변수들 : 식

lambda x : x + 10
```

```python
plus_ten2 = lambda x : x + 10

plus_ten2(10)
```

```python
# 람다표현식 자체를 호출하기

# (lambda 매개변수 : 식)(인수들)

(lambda x : x + 10)(1) 
```

```python
# 람다표현식 안에서는 변수를 만들수 없다.

(lambda x : y = 10 ; x + y)(1)
```

```python
# map에 객체를 여러개 넣기

a = [1, 2, 3, 4, 5]
b = [2, 4, 6, 8, 10]

list(map(lambda x, y : x * y, a, b))
```

```python
# filter
# filter(함수, 반복 가능한 객체)
## 필터 다음 무조건 함수 명만 적어야함 ex) f(x)가 아니라 f만 적어야함

def f(x):
    return x > 5 and x < 10

a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

list(filter(f, a))
```

```python
a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

list(filter(lambda x : x >5 and x < 10, a))
```

```python
# reduce 사용하기

from functools import reduce
# reduce(함수, 반복 가능한 객체)

def f(x, y):
    return x + y

a = [1, 2, 3, 4, 5]

reduce(f, a)
```

```python
# reduce에 람다 적용시키기

a = [1, 2, 3, 4, 5]

from functools import reduce

reduce(lambda x, y : x + y, a)
```
