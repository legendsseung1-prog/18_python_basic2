```python
# else 사용하기

# if 조건식:
#     수행할 코드
# else:
#     if를 제외한 모든 코드

x = 5
if x ==10:
    print("10입니다.")
else:
    print("10이 아닙니다.")
```

```python
# else와 들여쓰기
x = 5
if x ==10:
    print("10입니다.")
else:
    print("x에 들어있는 숫자는")
    print("10이 아닙니다.")
```

```python
# if 조건문의 동작 방식 알아보기

if True:
    print("참")
else:
    print("거짓")

if False:
    print("참")
else:
    print("거짓")
    
if None:
    print("참")
else:
    print("거짓")
```

```python
# 조건문에 숫자 지정하기

if 0:
    print("참")
else:
    print("거짓")
    
if 1:
    print("참")
else:
    print("거짓")
    
if 0x1f:
    print("참")
else:
    print("거짓")

if 0b1000:
    print("참")
else:
    print("거짓")
    
if 13.5:
    print("참")
else:
    print("거짓")
```

```python
# if조건문에 문자열 지정하기

if "hello":
    print("참")
else:
    print("거짓")

if "":
    print("참")
else:
    print("거짓")

if " ":
    print("참")
else:
    print("거짓")
```

```python
# 조건식을 여러개 지정하기

x = 10
y = 20

if x == 10 and y == 20:
    print("참")
else:
    print("거짓")
```

```python
# 중첩 if 조건문과 논리 연산자

x = 10

if x > 0:
    if x <20:
        print("20보다 작은 양수입니다.")
    
```

```python
if x > 0 and x < 20:
    print("20보다 작은 양수입니다.")
```

```python
written_test = 75
coding_test = True

if written_test >= 80 and coding_test == True:
    print("합격")
else:
    print("불합격")
```

```python
# 2가지 방법

# a = int(input("국어 : "))
# b = int(input("영어 : "))
# c = int(input("수학 : "))
# d = int(input("과학 : "))

a, b, c ,d = map(int,input("국어, 영어, 수학, 과학").split())

avg = (a + b + c + d) / 4

if 0 <= a <= 100 and 0 <= b <= 100 and 0 <= c <= 100 and 0 <= d <= 100:
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
