```python
# 덧셈함수 만들기

# def 함수이름(매개변수1, 매개변수2):
#     코드

def add(a, b):
    print(a + b)
    
result1 = add(10, 20)
print(result1)
```

```python
# 함수의 결과를 반환하기
# def 함수이름(매개변수1, 매개변수2):
#    retrun 매개변수를 사용한 코드

def add(a, b):
    return a + b
result2 = add(10, 20)
print(result2)
```

```python
#1 매개변수가 없고, 리턴되는 값도 없는 함수
#2 매개변수가 없고, 리턴되는 값이 있는 함수
#3 매개변수가 있고, 리턴되는 값이 있는 함수
#4 매개변수가 있고, 리턴되는 값이 없는 함수
```

```python
#1. 매개변수 x, 리턴값 x

def say_hello():
    print("hello")

say_hello()
```

```python
#2 매개변수 x, 리턴값은 o

def one():
    return 1
one()
```

```python
#3 매개변수 o, 리턴값은 x

def add(a, b):
    # a = 10, b = 20
    print(a + b)
```

```python
#4 매개변수 o, 리턴값은 o

def add(a, b):
    # a = 10, b = 20
    return(a + b)
```

```python
# return으로 함수 중간에서 빠져나오기

def not_ten(a):
    if a == 10:
        return
    print(a,"입니다.", sep = "")
```

```python
not_ten(10)
```

```python
# 함수에서 값을 여러개 반환하기

def add(a, b):
    return a + b
    return a - b

add(10, 20)
```

```python
def add(a, b):
    return a + b, a - b

result = add(10, 20)
result
```

```python
x, y = add(10, 20)
print(x)
print(y)
```

```python
# 함수의 호출 과정 알아보기

def mul(a, b):
    c = a * b
    return c
def add(a, b):
    c = a + b
    print(c)
    d = mul(a, b)
    print(d)
    
x = 10
y = 20
add(x, y)
```
