## 19.6 심사문제

```python
a = int(input())

for i in range(a):
    print(" " * (a - 1 - i), end = "")
    print("*" * (2 * i + 1))
```

```python
#2

a = int(input())

for i in range(a):
    for j in range(a - i - 1):
        print(" ", end="")
    for j in range(2 * i + 1):
        print("*", end="")
    print() 
```

## 20.7번 연습문제

```python
for i in range(1 , 101):
    if i % 2 == 0 and i % 11 == 0:
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```
