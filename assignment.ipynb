## 23.7번 심사문제 ★★★★★시험★★★★★

```python
# 1번 표준입력 : 3 3  .**  *..  .*.  
# 2번 표준입력 : 5 5  ..*..  ...*.  .*...  .***.  *.*..


# 1. 8방향 움직임 규칙을 맨 위에 먼저 딱 지정해!
# (세로변화, 가로변화) 순서야.
DIRECTIONS = [
    (-1, -1), (-1, 0), (-1, 1),  # 위 왼쪽,  위,  위 오른쪽
    (0, -1),           (0, 1),   # 왼쪽,          오른쪽
    (1, -1),  (1, 0),  (1, 1)    # 아래 왼쪽, 아래, 아래 오른쪽
]

# 2. 지도 크기 입력받기 (가로, 세로)
col, row = map(int, input().split())

# 3. 지도 입력받기 (지뢰는 *, 빈칸은 .)
matrix = []
for i in range(row):
    matrix.append(list(input()))

# 4. 지도 전체를 돌며 지뢰 개수 세기
for i in range(row):        # 세로 줄 번호 (i)
    for j in range(col):    # 가로 칸 번호 (j)

        # 지뢰가 있는 칸은 계산할 필요 없이 바로 * 출력!
        if matrix[i][j] == "*":
            print("*", end="")

        # 빈칸인 경우에만 주변 8방향을 검사해
        else:
            count = 0

            # 맨 위에 정해둔 8방향 규칙을 하나씩 꺼내서 더해봐
            for dy, dx in DIRECTIONS:
                ny = i + dy  # 체크할 주변 세로 위치
                nx = j + dx  # 체크할 주변 가로 위치

                # 체크할 위치가 지도 안쪽이면서, 그 자리가 지뢰(*)라면!
                if 0 <= ny < row and 0 <= nx < col:
                    if matrix[ny][nx] == "*":
                        count += 1

            # 다 셌으면 개수를 출력!
            print(count, end="")

    print()  # 한 줄 출력이 끝났으니 줄바꿈
```

```python
# 1번 표준입력 : 3 3  .**  *..  .*.  
# 2번 표준입력 : 5 5  ..*..  ...*.  .*...  .***.  *.*..

col, row = map(int, input().split())  # 가로 칸 수와 세로 줄 수를 입력받아.

matrix = []  # 지도를 담을 빈 리스트를 만들어.

for i in range(row):  # 세로 줄 수만큼 반복해.
    matrix.append(list(input()))  # 입력한 한 줄을 글자별로 나눠 지도에 넣어.

for i in range(row):  # 맨 위 줄부터 한 줄씩 살펴봐. i는 줄 번호야.
    for j in range(col):  # 왼쪽부터 한 칸씩 살펴봐. j는 칸 번호야.
        if matrix[i][j] == "*":  # 지금 칸에 지뢰가 있으면
            print("*", end="")  # *를 출력하고 줄은 바꾸지 않아.
        else:  # 지금 칸이 빈칸이면
            count = 0  # 주변 지뢰 개수를 0부터 세기 시작해.

            for y in range(i - 1, i + 2):  # 위쪽 줄, 현재 줄, 아래쪽 줄을 살펴봐.
                for x in range(j - 1, j + 2):  # 왼쪽 칸, 현재 칸, 오른쪽 칸을 살펴봐.
                    if 0 <= y < row and 0 <= x < col:  # 살펴볼 칸이 지도 안에 있으면
                        if matrix[y][x] == "*":  # 그 칸에 지뢰가 있는지 확인해.
                            count += 1  # 지뢰를 찾았으니 개수를 1 늘려.

            print(count, end="")  # 센 지뢰 개수를 출력하고 줄은 바꾸지 않아.

    print()  # 가로 한 줄을 다 출력했으니 다음 줄로 내려가.
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == ".":
            
            
            count = 0
            
            for r in range(i - 1, i + 2):
                for c in range(j - 1, j + 2):
                    if 0 <= r < row and 0 <= c = col:
                        if matrix[r][c] == "*":
                            count += 1
            
            matrix[i][j] = count
            
print(*matrix, sep = "\n")
```
