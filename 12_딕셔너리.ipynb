```python
a = input("이름을 적어주십시오.").split()
b = map(int,input("가산점을 적어주십시오.").split())

scores = {
          "승민" : 80,
          "학수" : 85,
          "태훈" : 90
}

for name, bonus in zip(a, b):
    scores[name] = scores[name] + bonus
print(scores)
```
