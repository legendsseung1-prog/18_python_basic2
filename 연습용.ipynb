```python
# 1. 8방향 움직임 규칙을 맨 위에 먼저 딱 지정해!
# (세로변화, 가로변화) 순서야.
DIRECTIONS = [
    (-1, -1), (-1, 0), (-1, 1),  # 위 왼쪽,  위,  위 오른쪽
    (0, -1),           (0, 1),   # 왼쪽,          오른쪽
    (1, -1),  (1, 0),  (1, 1)    # 아래 왼쪽, 아래, 아래 오른쪽
]

# 2. 지도 크기 입력받기 (가로, 세로)
col, row = map(int, input().split())

# 3. 지도 입력받기 (지뢰는 *, 빈칸은 .)
matrix = []
for i in range(row):
    matrix.append(list(input()))

# 4. 지도 전체를 돌며 지뢰 개수 세기
for i in range(row):        # 세로 줄 번호 (i)
    for j in range(col):    # 가로 칸 번호 (j)

        # 지뢰가 있는 칸은 계산할 필요 없이 바로 * 출력!
        if matrix[i][j] == "*":
            print("*", end="")

        # 빈칸인 경우에만 주변 8방향을 검사해
        else:
            count = 0

            # 맨 위에 정해둔 8방향 규칙을 하나씩 꺼내서 더해봐
            for dy, dx in DIRECTIONS:
                ny = i + dy  # 체크할 주변 세로 위치
                nx = j + dx  # 체크할 주변 가로 위치

                # 체크할 위치가 지도 안쪽이면서, 그 자리가 지뢰(*)라면!
                if 0 <= ny < row and 0 <= nx < col:
                    if matrix[ny][nx] == "*":
                        count += 1

            # 다 셌으면 개수를 출력!
            print(count, end="")

    print()  # 한 줄 출력이 끝났으니 줄바꿈
```

```python
records = []

def order():
    date = input()
    quantity = int(input())
    sales = quantity * 3000
    profit = quantity * 2000
    records.append([date, sales, profit])
    print("Total:", sales, "won")

def report(period):
    sales = 0
    profit = 0
    for date, amount, earnings in records:
        if date.startswith(period):
            sales += amount
            profit += earnings
    print("Sales:", sales, "won / Profit:", profit, "won")

order()
report("09-21")
report("09")
```

```python
기록 = []
가격 = [3000, 4000, 4500]
원가 = [1000, 1500, 2000]

def 주문():
    print("1. 아메리카노  2. 라떼  3. 모카")
    날짜 = input("날짜 (예: 09-21): ")
    메뉴 = int(input("메뉴 번호 (1~3): ")) - 1
    수량 = int(input("주문 수량: "))
    매출 = 가격[메뉴] * 수량
    수익 = 매출 - 원가[메뉴] * 수량
    기록.append([날짜, 매출, 수익])
    print("결제 금액:", 매출, "원")

def 일별():
    날짜 = input("조회할 날짜 (예: 09-21): ")
    매출 = 0
    수익 = 0
    for 항목 in 기록:
        if 항목[0] == 날짜:
            매출 += 항목[1]
            수익 += 항목[2]
    print("일별 매출:", 매출, "원 / 수익:", 수익, "원")

def 월별():
    월 = input("조회할 월 (예: 09): ")
    매출 = 0
    수익 = 0
    for 항목 in 기록:
        if 항목[0][:2] == 월:
            매출 += 항목[1]
            수익 += 항목[2]
    print("월별 매출:", 매출, "원 / 수익:", 수익, "원")

while True:
    선택 = input("1. 주문  2. 일별 조회  3. 월별 조회  4. 종료: ")
    if 선택 == "1":
        주문()
    elif 선택 == "2":
        일별()
    elif 선택 == "3":
        월별()
    elif 선택 == "4":
        break
```
