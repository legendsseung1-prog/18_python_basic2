```python
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end="")
    print()
    
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print(' ', end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j < i:
            print(" ", end="")    
        else:
            print("*", end="")
    print()    
```

```python
n = int(input())

for i in range(n):
    for j in range(n - i - 1):
        print(" ", end="")
    for j in range(2 * i + 1):
        print("*", end="")
    print() 
```

```python
a = int(input())

for i in range(a):
    print(" " * ((a - 1) - i), end="")
    print("*" * ((2 * i) + 1))
```
