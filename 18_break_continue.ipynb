```python
# break로 반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 10:
        break
```

```python
# for에서 break로 반복문 끝내기
for i in range(10000):
    print(i)
    if i == 10:
        break
```

```python
# continue로 코드 실행 건너뛰기
for i in range(100):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# while 반복문에서 continue로 코드 실행 건너뛰기

i = 0

while i < 100:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 입력한 횟수대로 반복하기

count = int(input())

i = 0
while True:
    print(i)
    i += 1
    if i == count:
        break
```

```python
count = int(input())

for i in range(count + 1):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 연습문제

i = 0
while True:
    if i % 10 != 3:
        i += 1
        continue
    if i > 73:
        break
    print(i, end=" ")
    i += 1
```

```python
# 심사문제

start, stop = map(int,input().split())
# 1     20
i = start

while True:
    if i % 10 == 3:
        i += 1
        continue
    if i > stop:
        break
    print(i, end=" ")
    i += 1
```
