# 문제 3



```python
library = {
    "문학": [
        {"제목": "1984", "저자": "조지 오웰", "재고": 3},
        {"제목": "위대한 개츠비", "저자": "F. 스콧 피츠제럴드", "재고": 0}
    ],
    "과학": [
        {"제목": "코스모스", "저자": "칼 세이건", "재고": 5},
        {"제목": "이기적 유전자", "저자": "리처드 도킨스", "재고": 2}
    ]
}

# 1. 재고가 0인 책의 제목과 분야를 모두 출력하세요.

for genre, books in library.items():
    for book in books:
        if book["재고"] == 0:
            print(f"제목: {book['제목']}, 분야: {genre}")


# 2. 전체 재고 수량의 합계를 출력하세요.

total = 0

for books in library.values():
    for book in books:
        total = total + book["재고"]

print(f"전체 재고 수량: {total}")
```

# 문제 4



```python
start, end = map(int, input().split())

for number in range(start, end + 1):
    if number % 20 == 0:
        print("FourFive", end=" ")

    elif number % 4 == 0:
        print("Four", end=" ")

    elif number % 5 == 0:
        print("Five", end=" ")

    else:
        print(number, end=" ")
```

# 문제 5



```python
a = int(input())

# 위쪽
for i in range(a):
    print(" " * (a - 1 - i), end="")
    print("*" * (2 * i + 1))

# 아래쪽
for i in range(a - 2, -1, -1):
    print(" " * (a - 1 - i), end="")
    print("*" * (2 * i + 1))
```

# 문제 6



```python
start, end = map(int, input().split())

# 1. 일반 for문 방식
result1 = []

for number in range(start, end + 1):
    if number % 2 == 1:
        result1.append(number ** 3)

print(result1)


# 2. 리스트 컴프리헨션 방식
result2 = [number ** 3 for number in range(start, end + 1) if number % 2 == 1]

print(result2)
```

# 문제 7



```python
salaries = input().split(",")

# 문자열을 숫자로 바꾸기
salaries = [int(salary) for salary in salaries]

# 큰 값부터 정렬하기
salaries.sort(reverse=True)

# 급여 출력하기
for salary in salaries:
    print(f"{salary:,}원")
```

# 문제 8



```python
matrix = []

# 3줄을 입력받아 2차원 리스트 만들기
for _ in range(3):
    row = list(map(int, input().split()))
    matrix.append(row)

total = 0

# 모든 위치를 확인하기
for i in range(3):
    for j in range(3):
        # 가운데 [1][1]이 아니면 더하기
        if i != 1 or j != 1:
            total = total + matrix[i][j]

print(total)
```

# 문제 9



```python
def analyze_scores(*args):
    total = 0

    # 모든 점수 더하기
    for score in args:
        total = total + score

    # 평균 구하기
    average = total / len(args)

    # 평균 이상인 점수 개수 세기
    count = 0

    for score in args:
        if score >= average:
            count = count + 1

    return average, count


scores = list(map(int, input().split()))

average, count = analyze_scores(*scores)

print(f"평균: {average} 평균 이상인 수의 개수: {count}")
```

# 문제 10



```python
inventory = {
    "노트북": 1500000,
    "마우스": 30000,
    "키보드": 50000,           
    "모니터": 300000
}


def checkout(**kwargs):
    total = 0

    for product, quantity in kwargs.items():
        if product in inventory:
            price = inventory[product] * quantity
            total = total + price

            print(f"{product} {quantity}개: {price:,}원")

        else:
            print(f"판매하지 않는 상품입니다: {product}")

    print(f"총 결제 금액: {total:,}원")


checkout(노트북=1, 마우스=2)

print()

checkout(모니터=2, 패드=1)
```
