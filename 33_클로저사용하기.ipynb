```python
# 변수의 사용 범위 알아보기

x = 10
def foo():
    print(x)

foo()
print(x)
```

```python
def foo():
    y = 10
    print(y)

foo()
print(y)
```

```python
# 함수 안에서 전역 변수 변경하기

x = 10
def foo():
    x = 20
    print(x)
    
foo()
print(x)
```

```python
# global 전역변수

x = 10

def foo():
    global x
    x = 20
    print(x)
    
foo()
print(x)
```

```python
def foo():
    global a
    a = 20
    print(a)
    
foo()
print(a)
```

```python
# 함수 안에서 함수 만들기

def print_hello():
    hello = "hello world"
    def print_message():
        print("hello")
        print_message()
    print(print_hello)
    
print_hello()
```

```python
# 지역변수 변경하기

def a():
    x = 10
    def b():
        x = 20
        
    b()
    print(x)
    
a()
```

```python
# nonlocal, wldurqustn

def a():
    x = 10
    def b():
        nonlocal x
        x = 20
        
    b()
    print(x)
    
a()
```

```python
# nonlocal이 변수를 찾는 순서

def a():
    x = 10
    y = 100
    def b():
        x = 20
        def c():
            nonlocal x
            nonlocal y
            x = y + 30
            y = y + 300
            print(x)
            print(y)
        
        c()
    b()   
a()
```

```python
# global로 전역변수 사용하기

x = 1
def a():
    x = 10
    def b():
        x = 20
        def c():
            global x
            x = x + 30
            print(x)
        (c)
    b()
a()
```

```python
# 클로저 사용하기

def calc():
    a = 3
    b = 5
    def mul_add(x):
        return a * x + b
    return mul_add

c = calc()
print(c(1), c(2), c(3), c(4))
```

```python
# 람다로 클로저 만들기

def calc():
    a = 3
    b = 5
    return lambda x : a * x + bc 

c = calc()
print(c())
```

```python
# 클로저의 지역변수 변경학

def calc():
    a = 3
    b = 5
    total  = 0
    
    def mul_add(x):
        nonlocal total
        total = total + a * x + b
        print(total)
        
    return mul_add
c = calc()
c(1)
```
