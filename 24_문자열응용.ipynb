```python
# 문자열 서식 지정자와 포매팅 사용하기
## 서식 지정자로 문자열 넣기
### &s & "문자열"

"I am %s" % "james"
```

```python
name = "maria"
"I am %s" % name
```

```python
# 서식 지정자로 숫자 넣기
# "%d" % 숫자

"I am %d" % 20

b = 10

"I am %d" % b
```

```python
# 서식 지정자로 소수점 표현하기
# "%f" % 숫자(실수)

"%f" % 2.3
```

```python
# "%.자리수f" % 숫자

print("%.3f" %2.3) 
print("%5.2f" %2.3) 
```

```python
# 서식 지정자로 문자열 정렬하기
# %길이s

"%10s" % "python" 
```

```python
# 왼쪽 정렬하기

"%-10s" % "python"
```

```python
# 서식 지정자로 문자열 안에 값 여러개 넣기
# "%d %s" % (숫자, 문자)

print("Today is %d %s" % (3, "April"))
```

```python
# format 매서드 사용하기

print("Hello, {0}".format("world"))
```

```python
# format 매서드로 값을 여러개 넣기
## 0번쨰 2번째 1번째 인덱스 가져오는것 처럼 가져오는 느낌

"Hello {0} {2} {1}".format("python", "script", 3.6)
```

```python
# format 매서드로 같은 값을 여러개 넣기

print("{0} {0} {1} {1}".format("python"," script"))
```

```python
# format 매서드에 인덱스 생략하기

"Hello {} {} {}".format("python", "script", 3.6)
```

```python
# format 매서드에서 인덱스 대신에 이름 지정하기

print("Hello {language} {version}".format(language = "Python", version = 3.6))
```
