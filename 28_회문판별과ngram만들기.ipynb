```python
# 회문 판별하기
## 반복문으로 문자 검사하기

word = input("단어를 입력하세요")

is_palindrome = True

for i in range(len(word) // 2):
    if word[i] != word[-1 - i]:
        is_palindrome = False
        
        break
print(is_palindrome)
```

```python
# 시퀀스 뒤집기로 문자 검사하기

word = input()

print(word == word[::-1])
```

```python
# 리스트와 reversed 사용하기

word = "level"

print(list(word))
print(list(reversed(word)))
```

```python
###########################################################
word = "level"
word == "".join(reversed("word"))
```

```python
# N-gram 만들기
## 반복문으로 N-gram 출력하기

text = "hello"

for i in range(len(text) - 1):
    print(text[i], text[i+1], sep = "")
```

```python
text = "this is python script"

word = text.split()

for i in range(len(word) - 1):
    print(word[i], word[i+1])
```

```python
# zip으로 2-gram 만들기

text = "hello"

two_gram = zip(text,text[1:])
list(two_gram)

for i in two_gram:
    print(i[0], i[1], sep = "")
```

```python
# zip과 리스트 표현식으로 N-gram 만들기

text = "hello"

[text[i:] for i in range(3)]
```

```python
# zip뒤에 별표는 자주쓰임
list(zip(*['hello', 'ello', 'llo']))
```
