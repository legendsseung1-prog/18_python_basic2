```python
# 값을 여러개 출력하기
print(1, 2, 3)
print("hello","python")
```

```python
## sep로 값 사이에 문자 넣기
## sep = " " 이게 디폴트
print(1, 2, 3)
print(1, 2, 3, sep = ",")
```

```python
print("hello", "python")
print("hello", "python", sep = "")
```

```python
print(1920, 1080, sep = " x ")
```

```python
# 줄바꿈 활용
print(1, 2, 3)
print(1, 2, 3, sep = "\n")
```

```python
print("1\n2\n3")
```

```python
## end 사용하기
print(1)
print(2)
print(3)
```

```python
print(1, end = "")
print(2, end = "")
print(3, end = "")
```

```python
print(1, end = " ")
print(2, end = " ")
print(3, end = " ")
```

```python
print("hello", "\n","python", sep = "")
```

```python
year = 2000
month = 10
day = 27
hour = 11
minute = 43
second = 59
print(year, month, day, sep = "/", end = " ")
print(hour, minute, second, sep = ":")
```

```python
a = 1999
b = 12
c = 31
d = 10
e = 37
f = 21
print(a, b, c, sep = "-", end = "T")
print(d, e, f, sep = ":")
```
